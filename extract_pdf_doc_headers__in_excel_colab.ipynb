```python
import fitz  # PyMuPDF for PDF parsing
import pandas as pd
import re
import os
from google.colab import files
```

```python
# Upload the PDF file
print("Upload your PDF file:")
uploaded = files.upload()

# Get the uploaded filename
pdf_file = list(uploaded.keys())[0]
print(f"Processing file: {pdf_file}")
```

```python
def extract_headers_from_pdf(pdf_path, output_excel="headers_output.xlsx"):
    headers = []
    doc = fitz.open(pdf_path)
    
    # Regular expression to match headers like "1.1", "1.1.1", etc.
    header_pattern = re.compile(r'^(\d+(\.\d+)+)\s+(.+)$')

    found_headers = False  # Flag to track when headers start

    for page in doc:
        text = page.get_text("text")
        for line in text.split("\n"):
            match = header_pattern.match(line.strip())
            
            if match:
                found_headers = True
            
            if found_headers and match:
                header_number, _, header_text = match.groups()
                headers.append((header_number, header_text))

    # If no headers found, add a message instead of failing
    if not headers:
        headers.append(("No Headers Found", "Check the PDF formatting"))

    # Convert to DataFrame and save to Excel
    df = pd.DataFrame(headers, columns=["Header Number", "Header Text"])
    df.to_excel(output_excel, index=False)
    
    print(f"✅ Headers extracted and saved to {output_excel}")
```

```python
# Run the extraction function
excel_output = "headers_output.xlsx"
extract_headers_from_pdf(pdf_file, excel_output)

# Provide a download link for the Excel file
files.download(excel_output)
print("📥 Click the link above to download the Excel file.")
```
